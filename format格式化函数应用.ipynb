```python
name = 'Zed A. Shaw'
age = 35
height = 74 #英寸
weight = 90 #磅
eyes = 'blue'
teeth = 'white'
hair = '棕色'

height = height * 2.54 #英寸转厘米
weight = weight * 0.4536 #磅转Kg

print(f"让我们来谈谈{name}") #f{}格式化字符串，将{}中变量名的值放到字符串中
print(f"他们的身高是{height}") #以上只适用于python3.6以上版本
print(f"他的体重是{weight}磅={weight*0.4536}Kg") #在输出中变换单位
print(f"实际上这不算太重")
print(f"他的牙齿通常是{teeth}的，这取决于他喝了多少咖啡")

total = age + height + weight

total = round(total) #round()函数可以四舍五入或者保留规定小数
print(f"如果我把{age} + {height} + {weight} 就得可以= {total}")
```
